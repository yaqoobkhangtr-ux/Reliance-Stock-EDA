### 🔒 Python Tuples (Quick Reference)

* **What It Is:** An ordered, **immutable** (unchangeable) sequence created with parentheses `()`. 
  * *Rule:* Access items with square brackets `tuple[0]`, but modifying items (`tuple[0] = x`) crashes with a `TypeError`.
  * *Inbuilt Methods (Only 2!):* `.count(value)` and `.index(value)`.
* **Where Used in Data Science:** Storing fixed data that must **never change accidentally** during pipeline execution (Stage 1 Ingestion & Stage 4 Modeling).

```python
# 1. Dataset Dimensions (df.shape)
shape = (1000, 5)  # (1000 rows, 5 columns)

# 2. Fixed Geographic Coordinates
location = (12.9716, 77.5946)

# Accessing works, but modifying fails:
print(shape[0])  # Output: 1000
# shape[0] = 500  ──► TypeError (Protected Data)

### 🧮 NumPy vs. Arrays vs. Vectorization

* **NumPy:** The C-powered **math library** for Python.
* **Array:** The continuous **grid container** (`np.array`) that holds data of a single type.
* **Vectorization:** Performing mathematical operations on an entire array at once, replacing slow `for` loops with fast batch processing.

```python
import numpy as np

# 1. Standard Python List (Requires explicit loop for math)
py_list = [10, 20, 30]
list_result = [x + 5 for x in py_list]

# 2. NumPy Array (Vectorized: No loop needed)
np_arr = np.array([10, 20, 30])
arr_result = np_arr + 5  # Returns: array([15, 25, 35])

### 🧮 Real-Life Data Science Applications of NumPy Arrays

1. **Machine Learning Feature Inputs:** ML algorithms require 2D numerical NumPy arrays (`X_train`, `y_train`) to calculate weights.
2. **Image Processing:** Images are stored as 3D NumPy arrays (Height × Width × RGB channels) where pixels are modified via array math.
3. **Pandas Engine:** Pandas DataFrames are built directly on top of NumPy arrays to perform rapid column filtering and transformations.
